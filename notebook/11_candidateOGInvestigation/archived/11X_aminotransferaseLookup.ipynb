## Aminotransferase gene lookup (collaborator request)

Moved here from `11_candidateOGInvestigation.ipynb` - a one-off, untracked lookup checking
where a collaborator-supplied aminotransferase gene list (`OGOI`, from
`data/ATgene_sorghum_maize_syntelog_orthogroup_mapping.tsv`) falls relative to the phylogenetic
mixed model results, the a-priori DE gene sets, and the HyPhy RELAX candidates - not part of the
core candidate-OG identification pipeline. Kept unparameterized, matching the archived-code
convention - depends on `mappingFileMerged`, `testResList`, `candidateOGList`,
`envResponsiveGenes`, `hyphy_drought`/`hyphy_wet` built earlier in the core notebook, so it
isn't runnable standalone (re-run the relevant core cells first if revisiting this).

```r
resSummary = sapply(testResList[1:3],function(x) {
    t(sapply(OGOI,function(a) {
        tmp = subset(x,OG%in%a)$p
        if(length(tmp)!=0) return(tmp)
        else return(NA)
        }))
})
```

```r
rownames(resSummary) = names(OGOI)
```

```r
fwrite(resSummary,"/workdir/sh2246/p_phyloGWAS/output/aminotranferaseGene.significance.txt",sep = "\t",row.names = T)
```
