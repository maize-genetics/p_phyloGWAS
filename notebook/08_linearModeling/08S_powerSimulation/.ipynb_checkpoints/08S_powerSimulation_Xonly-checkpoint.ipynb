```r
# =============================================================================
# Simulation-Based Power Analysis: BM Trait Evolution with Mean Shift
# Phylogenetic Mixed Models | Grass Family (Poaceae)
# =============================================================================
# DEPENDENCIES: ape, phytools, ggplot2, dplyr
# Install if needed:
#   install.packages(c("ape", "phytools", "ggplot2", "dplyr"))
# =============================================================================

library(ape)
library(phytools)
library(ggplot2)
library(dplyr)
library(asreml)
library(parallel)
```

```r
# =============================================================================
# 2. DEFINE TRANSITION BRANCHES
# =============================================================================
# Specify which NODES mark the origin of a shifted regime.
# In phytools/ape, internal nodes are numbered (Ntip+1) to (Ntip+Nnode).
# All descendants of these nodes will carry the trait shift.

# --- EDIT THIS SECTION with your known transition nodes ---
n_tips  <- Ntip(tree)
n_nodes <- Nnode(tree)

# Example: two independent transitions
# Use plotTree(tree, show.node.label=TRUE) to identify your node numbers
transition_nodes <- rjson::fromJSON(file = '/workdir/sh2246/p_phyloGWAS/output/powerSimulation/realTransition.json')[[1]]           # <-- replace with your actual nodes
```

```r
# Helper: get all tip descendants of a node
get_descendants <- function(tree, node) {
  all_desc <- phytools::getDescendants(tree, node)
  tips_only <- all_desc[all_desc <= Ntip(tree)]
  tree$tip.label[tips_only]
}

# Assign regime to each tip
tip_regimes <- setNames(rep("ancestral", Ntip(tree)), tree$tip.label)
for (nd in transition_nodes[[1]]) {
  desc_tips <- get_descendants(tree, nd)
  tip_regimes[desc_tips] <- paste0("forward_shifted_node", nd)
}
for (nd in transition_nodes[[2]]) {
  desc_tips <- get_descendants(tree, nd)
  tip_regimes[desc_tips] <- paste0("reverse_shifted_node", nd)
}

cat("Regime counts:\n")
print(table(tip_regimes))
```

```r
# =============================================================================
# 3. SIMULATION PARAMETERS
# =============================================================================
sigma2     <- 0.5    # BM rate (variance per unit branch length)
shift_mean <- 2.0    # magnitude of mean shift in derived regime(s)
                     # set negative for a decrease
n_sims     <- 500    # number of simulation replicates

# Optional: different shifts per transition node
# shift_per_node <- c("shifted_node52" = 2.0, "shifted_node61" = -1.5)
```

```r
# =============================================================================
# 4. SINGLE-REPLICATE SIMULATION FUNCTION
# =============================================================================
simulate_bm_shift <- function(tree, tip_regimes, sigma2, shift_mean, seed = NULL) {

  if (!is.null(seed)) set.seed(seed)

  # Simulate BM on the full tree (shared ancestral process)
  bm_trait <- fastBM(tree, sig2 = sigma2, internal = FALSE)

  # Add mean shift to tips in derived regimes
  forward_shifted <- grep('forward',tip_regimes)
  reverse_shifted <- grep('reverse',tip_regimes)
  bm_trait[forward_shifted] <- bm_trait[forward_shifted] + shift_mean
  bm_trait[reverse_shifted] <- bm_trait[reverse_shifted] - shift_mean

  return(bm_trait)
}

# Quick test
test_sim <- simulate_bm_shift(tree, tip_regimes, sigma2, shift_mean, seed = 1)
cat("\nTrait means by regime (single replicate):\n")
print(tapply(test_sim, tip_regimes, mean))
```

```r
# =============================================================================
# 5. simulation LOOP
# =============================================================================

run_simulations <- function(tree, tip_regimes, sigma2, shift_mean,
                                n_sims = 500) {
    sim_trait = c()
    for (i in seq_len(n_sims)) {
        trait <- simulate_bm_shift(tree, tip_regimes, sigma2, shift_mean, seed = i)
        sim_trait = cbind(sim_trait,trait)
    }
    sim_trait = data.frame(tree$tip.label,sim_trait)
    colnames(sim_trait) = c("assemblyID",paste("sim",seq_len(n_sims),sep = "_"))
    return(sim_trait)
}
```

```r
fit_model <- function(data, Kmat,responseVar,predictor) {
#   library(asreml)
    gc()
    asreml.options(verbose = FALSE,trace = F)
    Coeff = c()
    WaldPval = c()
    for (i in predictor){
        fullFM = as.formula(paste0(responseVar, "~", i ))
        model_full <- asreml(fixed = fullFM,
                             random = ~ vm(assemblyID, Kmat) , ai.sing = F, data = data)

        modelWald <- wald.asreml(model_full)
        if(model_full$converge) {
            Coeff <- c(Coeff,model_full$coefficients$fixed[2])
            WaldPval <- c(WaldPval,modelWald[2,4])
        }else {
            Coeff <- c(Coeff,NA)
            WaldPval <- c(WaldPval,NA)
        }
    }
    
#     out = c(Coeff,WaldPval)
#     names(out) = c(paste("Coeff",predictor,sep = "_"),
#                    paste("P",predictor,sep = "_"))
    return(WaldPval)
}
```

```r
run_power_analysis = function(dat, KMat, responseVar, 
                              tree,tip_regimes,sigma2,shift_mean,
                              n_sims = 500, alpha = 0.05) {
    simDat = run_simulations(tree,tip_regimes,sigma2,shift_mean,n_sims)
    testDat = merge(dat,simDat)
    p = fit_model(testDat,KMat,responseVar,colnames(simDat)[-1])
    power = sum(p < alpha)/n_sims
    return(power)
}
```

```r
run_power_analysis(dat,phyloKMat,"envPC_1",tree,tip_regimes,sigma2,shift_mean,n_sims = 10,alpha = 0.05)
```

```r
# =============================================================================
# 6. POWER CURVE: vary shift magnitude
# =============================================================================
shift_range <- seq(0, 2, by = 0.25)
power_curve <- mclapply(shift_range, function(s) {
  r <- run_power_analysis(dat,phyloKMat,"envPC_1",
                          tree, tip_regimes, sigma2,
                          shift_mean = s, n_sims = 100)  # fewer sims for speed
  return(r)
},mc.cores = 9)

power_df <- data.frame(shift = shift_range, power = unlist(power_curve))
```

```r
p_curve <- ggplot(power_df, aes(x = shift, y = power)) +
  geom_line(color = "#2E8B57", linewidth = 1.2) +
  geom_point(color = "#2E8B57", size = 2.5) +
  geom_hline(yintercept = 0.8, linetype = "dashed", color = "grey40") +
  annotate("text", x = max(shift_range) * 0.9, y = 0.82,
           label = "80% power", color = "grey40", size = 3.5) +
  labs(
    title    = "Power Curve: Detecting Trait Shift Along Transition Branches",
    subtitle = sprintf("BM rate σ² = %.2f | %d simulations per shift value", sigma2, 100),
    x        = "Mean Shift Magnitude",
    y        = "Estimated Power"
  ) +
  scale_y_continuous(limits = c(0, 1), labels = scales::percent) +
  theme_classic(base_size = 13)

print(p_curve)
# ggsave("power_curve.pdf", p_curve, width = 7, height = 5)
```
