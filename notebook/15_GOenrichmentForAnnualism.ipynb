```r
library(topGO)
```

```r
mappingFile = read.table("/workdir/sh2246/p_phyloGWAS/output/OGToPv_mapping.txt")
annualTopHits = read.table("/workdir/sh2246/p_phyloGWAS/output/dnds_topHits.txt",header = T)
annualTopHits2 = read.table("/workdir/sh2246/p_phyloGWAS/output/lifeHistory_topHits.txt",header = T)
annualTopHits3 = read.table("/workdir/sh2246/p_phyloGWAS/output/shared_dnds_lifehistory_topHits.txt",header = T)
topHits = mappingFile[mappingFile[,2]%in%annualTopHits[,1],1]
topHits2 = mappingFile[mappingFile[,2]%in%annualTopHits2[,1],1]
topHits3 = mappingFile[mappingFile[,2]%in%annualTopHits3[,1],1]
```

```r
GODB <- readMappings("/workdir/sh2246/p_panAndOGASR/data/Pv_GOTable.txt",IDsep = ",")
background = intersect(names(GODB),mappingFile[,1])
# background = names(GODB)

tmp=factor(as.integer(background%in%topHits))
names(tmp)=background
tgd1=new( "topGOdata", ontology="BP", allGenes = tmp, nodeSize=5,annot=annFUN.gene2GO, gene2GO = GODB)
resTopGO.classic=runTest(tgd1, algorithm = "classic", statistic = "Fisher")
resTopGO.weight01=runTest(tgd1, algorithm = "weight01", statistic = "Fisher")
GO_res_table=GenTable(tgd1,Fisher.classic = resTopGO.classic,Fisher.weight01=resTopGO.weight01,orderBy = "Fisher.weight01",ranksOf="Fisher.classic",topNodes=length(resTopGO.classic@score),numChar=100)

tmp=factor(as.integer(background%in%topHits2))
names(tmp)=background
tgd2=new( "topGOdata", ontology="BP", allGenes = tmp, nodeSize=5,annot=annFUN.gene2GO, gene2GO = GODB)
resTopGO.classic=runTest(tgd2, algorithm = "classic", statistic = "Fisher")
resTopGO.weight01=runTest(tgd2, algorithm = "weight01", statistic = "Fisher")
GO_res_table2=GenTable(tgd2,Fisher.classic = resTopGO.classic,Fisher.weight01=resTopGO.weight01,orderBy = "Fisher.weight01",ranksOf="Fisher.classic",topNodes=length(resTopGO.classic@score),numChar=100)

tmp=factor(as.integer(background%in%topHits3))
names(tmp)=background
tgd3=new( "topGOdata", ontology="BP", allGenes = tmp, nodeSize=5,annot=annFUN.gene2GO, gene2GO = GODB)
resTopGO.classic=runTest(tgd3, algorithm = "classic", statistic = "Fisher")
resTopGO.weight01=runTest(tgd3, algorithm = "weight01", statistic = "Fisher")
GO_res_table3=GenTable(tgd3,Fisher.classic = resTopGO.classic,Fisher.weight01=resTopGO.weight01,orderBy = "Fisher.weight01",ranksOf="Fisher.classic",topNodes=length(resTopGO.classic@score),numChar=100)
```

```r
GO_res_table$sigGenes = sapply(GO_res_table[,1],function(x) {
    paste(mappingFile[mappingFile[,1]%in%intersect(topHits,genesInTerm(tgd1,x)[[1]]),2],collapse = ",")
})

GO_res_table2$sigGenes = sapply(GO_res_table2[,1],function(x) {
    paste(mappingFile[mappingFile[,1]%in%intersect(topHits2,genesInTerm(tgd2,x)[[1]]),2],collapse = ",")
})
GO_res_table3$sigGenes = sapply(GO_res_table3[,1],function(x) {
    paste(mappingFile[mappingFile[,1]%in%intersect(topHits3,genesInTerm(tgd3,x)[[1]]),2],collapse = ",")
})
```

```r
write.table(GO_res_table,"/workdir/sh2246/p_phyloGWAS/output/GO_dnds_topHits.txt",sep = "\t")
write.table(GO_res_table2,"/workdir/sh2246/p_phyloGWAS/output/GO_lifeHistory_topHits.txt",sep = "\t")
write.table(GO_res_table3,"/workdir/sh2246/p_phyloGWAS/output/GO_shared_dnds_lifehistory_topHits.txt",sep = "\t")
```
