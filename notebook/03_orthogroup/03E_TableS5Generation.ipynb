```r
####Table S5####
PHYLOGWAS_ROOT <- Sys.getenv("PHYLOGWAS_ROOT", unset = "/workdir/sh2246/p_phyloGWAS")
```

```r
library(Biostrings)
```

```r
OG_filtered = read.table(file.path(PHYLOGWAS_ROOT, "output/poaceaeHelixerOG_filtered_20250331.txt"),header = F)[,1]
```

```r
queryFa = readAAStringSet(file.path(PHYLOGWAS_ROOT, "output/poaceaeHelixerOG_ancSeq_gapRemoved_v2_20240909.fa"))
```

```r
OG_tested = read.table(file.path(PHYLOGWAS_ROOT, "output/allOG.txt"))[,1]
```

```r
angio353 = read.table(file.path(PHYLOGWAS_ROOT, "output/angiosperm353ToOGName.txt"),header = F)[,1]
```

```r
mappingFile = read.table(file.path(PHYLOGWAS_ROOT, "data/DEG_mappingFiles/OG_mapping_Oryza_sativa.IRGSP-1.0.cds.all.txt"))
mappingFile2 = read.table(file.path(PHYLOGWAS_ROOT, "data/OGToZm_mapping_v2.txt"))

colnames(mappingFile) = c("OsID","OG")
colnames(mappingFile2) = c("ZmID","OG")

mappingFileMerged = merge(mappingFile,mappingFile2,by = "OG",all = T)
```

```r
outTab = data.frame(OG = OG_filtered,ancAASeq = stringr::str_to_upper(as.character(queryFa)[OG_filtered]),
                    angiosperm353 = OG_filtered%in%angio353 ,tested = OG_filtered%in%OG_tested)
```

```r
outTab = merge(outTab,mappingFileMerged,by = "OG",all = T)
```

```r
outTab = outTab[,c(1,3,4,5,6,2)]
```

```r
write.table(outTab,file.path(PHYLOGWAS_ROOT, "output/TableS5.txt"),quote = F,sep = "\t",row.names = F)
```
