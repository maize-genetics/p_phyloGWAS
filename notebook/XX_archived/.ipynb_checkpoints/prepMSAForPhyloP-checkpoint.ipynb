```r
rm(list=ls())

library(ape)
library(pheatmap)
library(limma)
library(parallel)
library(tidyverse)
library(Biostrings)
```

```r
fasta.list=list.files("/workdir/sh2246/p_phyloGWAS_archived/output/CDSMSAPerOG_gs",".gs.l0.fa$",full.names = T)
OGNames = gsub(".gs.l0.fa","",strsplit2(fasta.list,"/")[,7])
fasta.aln=mclapply(fasta.list,function(x) readDNAStringSet(x),mc.cores = 35)
```

```r
metadata = read.delim("/workdir/sh2246/p_phyloGWAS/data/Poaceae_metadata_filtered_2025.04.16.tsv",header = T)
```

```r
length(metadata$assemblyID)
```

```r
subsetID = c("ASM1935983v1","IRGSP-1.0","Sorghum_bicolor_NCBIv3",setdiff(metadata$assemblyID[(metadata$complete+metadata$duplicated)>5200],c("ASM1935983v1","IRGSP-1.0","Sorghum_bicolor_NCBIv3")))
```

```r
length(subsetID)
```

```r
fasta.aln[[1]][1,]
```

```r
fasta.aln.processed = mclapply(fasta.aln,function(x){
    tmpM = x
    tmpM = tmpM[grep(':0$',names(tmpM)),]
    names(tmpM) = strsplit2(names(tmpM),":")[,2]

    missingTaxa = setdiff(subsetID,names(tmpM))

    newSeq = rep(paste(rep("-",width(tmpM[1,])),collapse = ""),length(missingTaxa))
    names(newSeq) = missingTaxa
    newSeq = DNAStringSet(newSeq)

    tmpM = c(tmpM,newSeq)
    tmpM = tmpM[subsetID,]
    return(tmpM)
},mc.cores = 20)
```

```r
names(fasta.aln.processed) = OGNames
```

```r
dir.create("/workdir/sh2246/p_phyloGWAS_archived/output/CDSMSAPerOG_phyloP_20250609/")
```

```r
mclapply(OGNames,function(x) {
    writeXStringSet(fasta.aln.processed[[x]],
                    filepath = paste0("/workdir/sh2246/p_phyloGWAS_archived/output/CDSMSAPerOG_phyloP_20250609/",x,".fa"),
                    format = "fasta")
},mc.cores = 20)
```

```r
tre = read.tree("/workdir/sh2246/p_phyloGWAS/output/PoaceaeTree_angiosperm353_astral_filtered_20250407.nwk")
```

```r
subtre = keep.tip(tre,subsetID)
```

```r
write.tree(subtre,"/workdir/sh2246/p_phyloGWAS_archived/output/PoaceaeTree_angiosperm353_astral_subsetForPhyloP_20250610.nwk")
```
