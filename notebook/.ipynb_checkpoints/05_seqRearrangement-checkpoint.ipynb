```r
rm(list=ls())
library(Biostrings)
library(parallel)
library(limma)
library(rtracklayer)
```

```r
faPerTaxa_flist = list.files("/workdir/hackathon_202309/data/miniprotPrimaryAlignmentVsPaspalum_filtered70identity/",pattern = "*.fa",full.names = T)
faPerTaxa_list = mclapply(faPerTaxa_flist,function(x) readDNAStringSet(x),mc.cores = 10)
```

```r
names(faPerTaxa_list)=gsub(".fa","",strsplit2(faPerTaxa_flist,"/")[,7])
```

```r
names(faPerTaxa_list[[1]])
```

```r
gff_flist = list.files("/workdir/hackathon_202309/data/miniprotPrimaryAlignmentVsPaspalum_filtered70identity/",pattern = "*.gff",full.names = T)
gff_list = mclapply(gff_flist,function(x) readGFF(x),mc.cores = 10)
```

```r
for (i in 1:length(gff_list)){
    print(i)
    idx = which(paste0(gff_list[[i]]$seqid,":",gff_list[[i]]$start-1,"-",gff_list[[i]]$end,"(",gff_list[[i]]$strand,")")%in%names(faPerTaxa_list[[i]]))
    names(faPerTaxa_list[[i]]) = strsplit2(gff_list[[i]]$Target," ")[idx,1]
}
```

```r
# OGID=read.table("/workdir/sh2246/p_panAndOGASR/output/OG_filtered_v2.txt",header = F)[,1]
OGID=read.table("/workdir/sh2246/p_evolBNI/output/Pv-672_allTranscriptID.txt",header = F)[,1]
```

```r
pvCDS = readDNAStringSet("/workdir/sh2246/p_evolBNI/output/Pv-672_v3.0_cds.fa")
```

```r
N = length(gff_list)+1
test=DNAStringSet(lapply(faPerTaxa_list,function(x) {
    tmp=x[["Pavag01G000100.1"]]
    if (is.null(tmp)) tmp = DNAString("-")
    return(tmp)
}))
test[[N]]=pvCDS[["Pavag01G000100.1"]]
names(test)[N]="Pavag01G000100.1"
test
```

```r
N = length(gff_list)+1

faPerOG_list = list()
for (i in OGID){
    faPerOG_list[[i]] = DNAStringSet(lapply(faPerTaxa_list,function(x) {
        tmp=x[[i]]
        if (is.null(tmp)) tmp = DNAString("-")
        return(tmp)
    }))
    faPerOG_list[[i]][[N]]=pvCDS[[i]]
    names(faPerOG_list[[i]])[N] = i
}
```
