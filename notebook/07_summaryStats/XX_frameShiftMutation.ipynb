```r
FSSeq = unlist(mclapply(gffFlist,function(x){
    gffTest = readGFF(x)
    tmp = gffTest[gffTest$type %in% "mRNA",]
    return(paste(tmp$ID[!is.na(tmp$Frameshift)],gsub('.gff','',strsplit2(x,"/")[,7]),sep = ":"))
},mc.cores = 20))
```
